```python
import pandas as pd
import numpy as np
```

```python
def cpi_reader(cpi):
    cpi = pd.read_excel(cpi,header=3)
    cpi.iloc[0,0:3] = cpi.columns[:3]
    cpi.columns = cpi.iloc[0].str.replace('.','').str.replace('\n',' ')
    cpi = cpi[cpi['Indent Level'].notna()]
    cpi = cpi.iloc[3:]
    cpi['Expenditure category'] = cpi['Expenditure category'].str.split('(').str[0]
    cpi = cpi.iloc[:,[0,1]+list(range(16,29))]
    cpi = cpi.set_index('Expenditure category').T
    cpi = cpi.replace('-', np.nan).apply(pd.to_numeric, errors='coerce')
    return cpi
```

```python
print(cpi_reader(r'CPI-U-202212.xlsx'))
```
